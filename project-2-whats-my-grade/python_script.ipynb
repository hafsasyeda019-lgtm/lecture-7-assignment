```
# ==========================================
# Grade Calculator
# Calculates the current grade based on
# assignments, quizzes, and midterm.
# Also calculates the required final exam
# score to achieve a target overall grade.
# ==========================================

# -----------------------------
# Student Scores
# -----------------------------
assignments = [88, 94, 91, 85]
quizzes = [9, 10, 8, 7, 9]
midterm = 82

# -----------------------------
# Grading Weights
# -----------------------------
assignment_weight = 0.25
quiz_weight = 0.20
midterm_weight = 0.25
final_weight = 0.30

# -----------------------------
# Assignment Average
# -----------------------------
assignment_average = sum(assignments) / len(assignments)

# -----------------------------
# Quiz Average
# Drop the lowest quiz score
# -----------------------------
lowest_quiz = min(quizzes)
quizzes.remove(lowest_quiz)
quiz_average = sum(quizzes) / len(quizzes)

# Convert quiz average from /10 to percentage
quiz_percentage = quiz_average * 10

# -----------------------------
# Current Grade
# (Final Exam not taken yet)
# -----------------------------
current_points = (
    assignment_average * assignment_weight +
    quiz_percentage * quiz_weight +
    midterm * midterm_weight
)

completed_weight = (
    assignment_weight +
    quiz_weight +
    midterm_weight
)

current_grade = (current_points / completed_weight)

# -----------------------------
# Display Current Results
# -----------------------------
print("=" * 50)
print("CURRENT GRADE REPORT")
print("=" * 50)

print(f"Assignment Average : {assignment_average:.2f}%")
print(f"Lowest Quiz Dropped: {lowest_quiz}")
print(f"Quiz Average       : {quiz_average:.2f}/10")
print(f"Quiz Percentage    : {quiz_percentage:.2f}%")
print(f"Midterm Score      : {midterm}%")
print(f"\nCurrent Grade      : {current_grade:.2f}%")

print("=" * 50)

# -----------------------------
# Target Grade
# -----------------------------
target = float(input("Enter your target overall grade (%): "))

# -----------------------------
# Calculate Required Final Score
# -----------------------------

required_final = (
    target
    - (
        assignment_average * assignment_weight
        + quiz_percentage * quiz_weight
        + midterm * midterm_weight
    )
) / final_weight

print("\n========== FINAL EXAM TARGET ==========")

if required_final <= 100:

    if required_final <= midterm:
        print(f"You need at least {required_final:.2f}% on the Final Exam.")

    else:
        # Special Rule:
        # Final replaces Midterm if Final > Midterm

        required_final = (
            target
            - (
                assignment_average * assignment_weight
                + quiz_percentage * quiz_weight
            )
        ) / (midterm_weight + final_weight)

        print("Special Rule Applied:")
        print("If your Final Exam score is higher than your Midterm,")
        print("it replaces the Midterm score.")

        print(f"\nYou need at least {required_final:.2f}% on the Final Exam.")

else:
    print("It is not possible to reach the target grade.")
```
