```
import pandas as pd
from google.colab import files

# Upload CSV file
uploaded = files.upload()

# Read CSV
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

# Verified total
verified_total = 6000

# Trip-related memos
trip_memos = {
    "trip",
    "trip fee",
    "class trip",
    "classtrip",
    "trip2026",
    "trp"
}

# Replace abbreviated names
df["Sender Name"] = df["Sender Name"].replace({
    "A. Ali": "Ahmed Ali"
})

# Function to identify trip payments
def is_trip_payment(row):
    memo = str(row["Memo"]).strip().lower()

    if memo == "nan":
        memo = ""

    if memo in trip_memos:
        return True

    if memo == "" and row["Amount (PKR)"] == 500:
        return True

    return False

# Apply rules
df["Trip Payment"] = df.apply(is_trip_payment, axis=1)

# Filter trip payments
trip_df = df[df["Trip Payment"]]

# Calculate totals
calculated_total = trip_df["Amount (PKR)"].sum()
difference = verified_total - calculated_total

# Follow-up payments
follow_up = df[~df["Trip Payment"]]

print("=" * 50)
print("BOOK RECONCILIATION REPORT")
print("=" * 50)

print(f"Verified Total : Rs. {verified_total}")
print(f"Calculated Total : Rs. {calculated_total}")
print(f"Difference : Rs. {difference}")

print("\nTrip Payments:")
print(trip_df[["Sender Name", "Amount (PKR)", "Memo"]])

print("\nPayments Requiring Follow-Up:")
if follow_up.empty:
    print("No unmatched payments found.")
else:
    print(follow_up[["Sender Name", "Amount (PKR)", "Memo"]])

print("\nSummary")
if difference == 0:
    print("Books successfully reconciled.")
else:
    print(f"Rs. {difference} is unaccounted for.")
    print("Follow up with the people listed above.")
```
