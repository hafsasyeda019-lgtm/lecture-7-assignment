```
import pandas as pd

# Load the transaction data
df = pd.read_csv("transaction.csv")

# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"])

# -----------------------------
# 1. Total Spending
# -----------------------------
total_spending = df["Amount"].sum()
print("=" * 50)
print(f"Total Spending: Rs. {total_spending:.2f}")

# -----------------------------
# 2. Monthly Totals
# -----------------------------
df["Month"] = df["Date"].dt.to_period("M")

monthly_totals = df.groupby("Month")["Amount"].sum()

print("\nMonthly Totals")
print("-" * 50)
print(monthly_totals)

# -----------------------------
# 3. Spending by Category
# -----------------------------
category_totals = df.groupby("Category")["Amount"].sum().sort_values(ascending=False)

print("\nSpending by Category")
print("-" * 50)
print(category_totals)

# Biggest spending category
biggest_category = category_totals.idxmax()
biggest_amount = category_totals.max()

print(f"\nBiggest Category: {biggest_category} (Rs. {biggest_amount:.2f})")

# -----------------------------
# 4. Recurring Payments
# -----------------------------
print("\nRecurring Payments")
print("-" * 50)

recurring = (
    df.groupby("Description")
      .agg(
          Count=("Description", "count"),
          Total=("Amount", "sum"),
          Amount=("Amount", "first")
      )
)

recurring = recurring[recurring["Count"] > 1]

if recurring.empty:
    print("No recurring payments found.")
else:
    print(recurring)

# -----------------------------
# 5. Duplicate Transactions
# -----------------------------
print("\nDuplicate Transactions")
print("-" * 50)

duplicates = df[df.duplicated(subset=["Date", "Description", "Amount"], keep=False)]

if duplicates.empty:
    print("No duplicate transactions found.")
else:
    print(duplicates.sort_values(["Date", "Description"]))

print("\nAnalysis Complete!")
```
